````python
import numpy as np
from itertools import combinations

class PrimeEntanglementGraph:
    """
    Represents a quantum state using a prime-factorization network rather 
    than a dense 2^N tensor. Entanglement between particles is represented 
    by shared prime divisors.
    """
    def __init__(self, num_particles):
        self.num_particles = num_particles
        # Each particle's state is tracked by the prime factors it contains.
        self.particle_factors = {i: [] for i in range(num_particles)}
        
        # Maps a prime divisor to its correlation weight (stability/coherence)
        self.bond_weights = {} 
        self.current_prime_idx = 0
        self.primes = self._generate_primes(50000)

    def _generate_primes(self, limit):
        """Generates a list of prime numbers using a simple sieve."""
        sieve = [True] * limit
        for p in range(2, int(limit**0.5) + 1):
            if sieve[p]:
                for i in range(p * p, limit, p):
                    sieve[i] = False
        return [p for p in range(2, limit) if sieve[p]]

    def entangle(self, p1, p2, coherence_weight=1.0):
        """
        Entangles two particles by assigning them a shared prime divisor.
        """
        shared_prime = self.primes[self.current_prime_idx]
        self.current_prime_idx += 1
        
        self.particle_factors[p1].append(shared_prime)
        self.particle_factors[p2].append(shared_prime)
        self.bond_weights[shared_prime] = coherence_weight

    def get_dense_tensor_size(self):
        """Returns the parameter count for a standard dense 2^N state vector."""
        return 2 ** self.num_particles

    def get_graph_size(self):
        """Returns the parameter count for the prime-graph representation."""
        # Size = Number of particles + Number of active shared prime bonds
        return self.num_particles + len(self.bond_weights)

    def apply_stability_filter(self, threshold):
        """
        Filters out highly chaotic or unstable entanglement bonds.
        If a shared prime bond has a coherence weight below the threshold,
        it is pruned from the graph.
        """
        primes_to_remove = []
        for prime, weight in self.bond_weights.items():
            if abs(weight) < threshold:
                primes_to_remove.append(prime)
                
        # Remove the unstable primes from the particles and the bond dictionary
        for prime in primes_to_remove:
            del self.bond_weights[prime]
            for p_id in self.particle_factors:
                if prime in self.particle_factors[p_id]:
                    self.particle_factors[p_id].remove(prime)
                    
        return len(primes_to_remove)

# ==========================================
# Simulation Runner
# ==========================================

def run_simulation():
    N = 20 # Number of particles
    print(f"--- SIMULATING N={N} PARTICLES ---")
    print(f"Standard Dense Tensor Size: {2**N:,} parameters\n")

    # ---------------------------------------------------------
    # Scenario 1: Area-Law (Structured Ground State)
    # ---------------------------------------------------------
    area_law_graph = PrimeEntanglementGraph(N)
    # Particles only entangle with their immediate neighbors (1D Spin Chain)
    for i in range(N - 1):
        area_law_graph.entangle(i, i + 1, coherence_weight=1.0)
        
    print("[1] AREA-LAW ENTANGLEMENT (Structured)")
    print(f"    Prime Graph Size: {area_law_graph.get_graph_size():,} parameters")
    print(f"    Compression Ratio: {(2**N) / area_law_graph.get_graph_size():,.1f}x\n")

    # ---------------------------------------------------------
    # Scenario 2: Chaotic (Dense All-to-All Entanglement)
    # ---------------------------------------------------------
    np.random.seed(42)
    chaotic_graph = PrimeEntanglementGraph(N)
    # Every particle entangles with every other particle with varying stability
    for p1, p2 in combinations(range(N), 2):
        # Generate a noisy correlation weight (e.g., thermal noise or chaotic scattering)
        noisy_weight = np.random.normal(loc=0.0, scale=0.5)
        chaotic_graph.entangle(p1, p2, coherence_weight=noisy_weight)

    # Note: A true Haar-random state would require hyper-edges scaling at 2^N.
    # We use all-to-all dense pairs here to illustrate the topological explosion.
    print("[2] CHAOTIC ENTANGLEMENT (Dense Noise)")
    print(f"    Prime Graph Size: {chaotic_graph.get_graph_size():,} parameters")
    print(f"    Compression Ratio: {(2**N) / chaotic_graph.get_graph_size():,.1f}x (Graph explodes)\n")

    # ---------------------------------------------------------
    # Scenario 3: Applying the Stability Filter
    # ---------------------------------------------------------
    # We filter out any entanglement bond that lacks the coherent "energy" 
    # to maintain structural stability.
    stability_threshold = 0.85 
    dropped_bonds = chaotic_graph.apply_stability_filter(stability_threshold)
    
    print("[3] APPLYING STABILITY FILTER TO CHAOS")
    print(f"    Threshold set to: {stability_threshold}")
    print(f"    Unstable bonds pruned: {dropped_bonds}")
    print(f"    Filtered Graph Size: {chaotic_graph.get_graph_size():,} parameters")
    print(f"    Recovered Compression: {(2**N) / chaotic_graph.get_graph_size():,.1f}x")

if __name__ == "__main__":
    run_simulation()
````

````python
import numpy as np
import scipy.linalg
from typing import List, Dict, Tuple

class ArithmeticTensorOracle:
    """
    The Oracle: Evaluates atomic configurations by mapping their spatial 
    and electronic correlations into an Arithmetic Fock Space, then filtering 
    out high-energy unstable noise to find the stable ground state.
    """
    def __init__(self, max_primes: int = 1000):
        self.primes = self._sieve(max_primes)
        
    def _sieve(self, n: int) -> List[int]:
        sieve = [True] * n
        for p in range(2, int(n**0.5) + 1):
            if sieve[p]:
                for i in range(p*p, n, p):
                    sieve[i] = False
        return [p for p in range(2, n) if sieve[p]]

    def _map_to_fock_exponents(self, intensity: float, seed_idx: int) -> List[int]:
        """
        Maps a continuous interaction intensity into a discrete set of prime exponents 
        (p-adic valuations) inside the arithmetic vacuum.
        """
        # Deterministic but complex mapping to simulate discrete atomic/electronic modes
        np.random.seed(seed_idx + int(intensity * 1000) % 1000)
        num_active_primes = min(int(intensity * 8) + 1, 10)
        
        exponents = [0] * len(self.primes)
        chosen_indices = np.random.choice(len(self.primes) // 4, num_active_primes, replace=False)
        for idx in chosen_indices:
            exponents[idx] = int(np.random.randint(1, 4))
        return exponents

    def evaluate_lattice(self, atom_positions: np.ndarray, stability_threshold: float) -> Dict[str, float]:
        """
        Takes an atomic configuration, builds the raw correlation hypergraph,
        applies the Obstruction-Filtered Positivity Sheaf, and extracts metrics.
        """
        num_atoms = len(atom_positions)
        
        # 1. Compute raw geometric distances to initialize the background metric H_0
        # Close atoms interact more strongly; distant atoms are decoupled.
        dist_matrix = np.linalg.norm(atom_positions[:, None, :] - atom_positions[None, :, :], axis=-1)
        # Avoid division by zero on self-interaction
        np.fill_diagonal(dist_matrix, 1.0) 
        
        # H_0 represents the unentangled continuous background metric.
        # We use a Gaussian RBF kernel to guarantee the matrix is positive-definite,
        # which is required to solve the generalized eigenvalue problem.
        H_0 = np.exp(-dist_matrix**2 / 2.0)
        np.fill_diagonal(H_0, 1.0)
        # Add a tiny ridge regularization to ensure strict numerical stability
        H_0 += np.eye(num_atoms) * 1e-5
        
        # 2. Map interactions to the Arithmetic Fock Space (Dense 'Hairball' State)
        # We assign arithmetic valuations based on neighbor proximity
        atom_exponents = []
        for i in range(num_atoms):
            local_density = np.sum(H_0[i]) - 1.0
            atom_exponents.append(self._map_to_fock_exponents(local_density, i))
            
        # Compute the raw interaction matrix H_int via shared prime divisors (GCD)
        H_int = np.zeros((num_atoms, num_atoms))
        raw_bonds_count = 0
        
        for i in range(num_atoms):
            for j in range(i + 1, num_atoms):
                # E_ij = \sum min(v_p(i), v_p(j)) * log(p)
                shared_log_valuation = 0.0
                for p_idx, p in enumerate(self.primes):
                    v_p_i = atom_exponents[i][p_idx]
                    v_p_j = atom_exponents[j][p_idx]
                    if v_p_i > 0 and v_p_j > 0:
                        shared_log_valuation += min(v_p_i, v_p_j) * np.log(p)
                
                if shared_log_valuation > 0:
                    H_int[i, j] = shared_log_valuation
                    H_int[j, i] = shared_log_valuation
                    raw_bonds_count += 1

        # 3. Apply Obstruction-Filtered Positivity Sheaf (The Filter)
        # Compute the contractive stability matrix: B = H_0^(-1/2) * H_int * H_0^(-1/2)
        # To avoid matrix inversion instabilities, we solve via generalized eigenvalues
        
        # We must apply a coupling constant to normalize the arithmetic Fock space 
        # so it acts as a proper geometric perturbation relative to the H_0 metric.
        coupling_constant = 0.05 
        H_int_scaled = H_int * coupling_constant    
        try:
            # eigh solves the generalized eigenvalue problem natively
            eigenvalues = scipy.linalg.eigh(H_int_scaled, H_0, eigvals_only=True)
            max_eigenvalue = np.max(np.abs(eigenvalues))
        except (np.linalg.LinAlgError, scipy.linalg.LinAlgError, ValueError):
            # Fallback if H_0 becomes computationally singular or degenerate
            max_eigenvalue = 10.0
            
        # If max_eigenvalue > 1.0, the Cone-Contractive Criterion (B <= I) is violated.
        # The filter dynamically prunes bonds proportional to the instability scaling.
        stable_bonds_count = raw_bonds_count
        if max_eigenvalue > 1.0:
            # Drop weak or highly unstable erratic bonds matching the filter threshold
            prune_probability = 1.0 - (1.0 / max_eigenvalue)
            stable_bonds_count = int(raw_bonds_count * (1.0 - (prune_probability * stability_threshold)))

        # 4. Measure Superconducting Properties (The Reward Signal)
        # High-temperature superconductivity thrives on clean, symmetric, highly-ordered 2D grids.
        # We calculate the pairing correlation length based on the remaining stable manifold.
        if stable_bonds_count == 0:
            sc_score = 0.0
        else:
            # Cooper-pair efficiency tracks uniform, localized coordination.
            # We add 1.0 to the denominator to prevent asymptotic explosions 
            # if the lattice accidentally achieves perfect uniformity prematurely.
            coordination_uniformity = 1.0 / (np.std(np.sum(H_int > 0, axis=1)) + 1.0)
            sc_score = (stable_bonds_count / (raw_bonds_count + 1e-5)) * coordination_uniformity * 50.0
            
        return {
            "raw_bonds": raw_bonds_count,
            "stable_bonds": stable_bonds_count,
            "max_instability": max_eigenvalue,
            "superconducting_score": min(sc_score, 400.0) # Cap at absolute theoretical limit (400K)
        }


class LatticeGenerator:
    """
    The Generator: Generates and iteratively mutates atomic crystal coordinates 
    using a pseudo-gradient ascent tracking the oracle's reward signal.
    """
    def __init__(self, num_atoms: int = 16, dimensions: int = 3):
        self.num_atoms = num_atoms
        self.dimensions = dimensions
        # Initialize a random cloud of atoms
        self.current_lattice = np.random.uniform(-2.0, 2.0, (num_atoms, dimensions))
        
    def propose_mutation(self, temperature: float) -> np.ndarray:
            """Adds thermal perturbation and applies a binding pressure to maintain a solid lattice."""
            mutation = np.random.normal(0.0, temperature, self.current_lattice.shape)
            proposed = self.current_lattice + mutation
            
            # Crystal Confinement: Pull atoms slightly towards the center of mass
            # This prevents the AI from "solving" instability by just blowing the atoms apart
            center_of_mass = np.mean(proposed, axis=0)
            proposed = proposed - 0.05 * (proposed - center_of_mass)
            
            return proposed

    def update_lattice(self, accepted_lattice: np.ndarray):
        self.current_lattice = accepted_lattice.copy()


# =========================================================================
# Closed Discovery Loop Execution
# =========================================================================

def run_discovery_loop(epochs: int = 100):
    oracle = ArithmeticTensorOracle()
    generator = LatticeGenerator(num_atoms=12, dimensions=3)
    
    best_score = 0.0
    best_lattice = None
    
    print("Initializing The Divergence Lab Material Search Loop...")
    print(f"Targeting: Ground-State Matrix Stability under Filter Threshold = 0.85\n")
    print(f"{'Epoch':<8}{'Raw Bonds':<12}{'Stable Bonds':<14}{'Instability':<14}{'Est. Tc (K)':<12}")
    print("-" * 65)
    
    # Annealing schedule for our generator's structural adjustments
    mutation_temp = 0.5

    for epoch in range(1, epochs + 1):
        # Step 1: Generator proposes an atomic configuration
        proposed_lattice = generator.propose_mutation(temperature=mutation_temp)
        
        # Step 2 & 3: Oracle maps to Fock space and applies the Positivity Sheaf filter
        metrics = oracle.evaluate_lattice(proposed_lattice, stability_threshold=0.85)
        
        raw_b = metrics["raw_bonds"]
        stable_b = metrics["stable_bonds"]
        instability = metrics["max_instability"]
        score = metrics["superconducting_score"]
        
        # Step 4: Process reward signal and steer the generation matrix
        # If the mutation yielded a higher pairing score, update the generator's baseline
        if score > best_score or (score > 0 and np.random.rand() < 0.05):
            best_score = score
            best_lattice = proposed_lattice
            generator.update_lattice(proposed_lattice)
            
        # Decay mutation variance over time to lock onto local minima/stable crystals
        mutation_temp = max(0.05, mutation_temp * 0.98)
        
        if epoch % 10 == 0 or epoch == 1:
            print(f"{epoch:<8}{raw_b:<12}{stable_b:<14}{instability:<14.3f}{score:<12.1f}")
            
    print("-" * 65)
    print(f"Discovery phase finished.")
    print(f"Highest Predicted Superconducting Transition Temperature: {best_score:.2f} Kelvin")
    print("\nOptimized Atomic Geometry (First 3 Coordinate Vectors):")
    print(best_lattice[:3])

if __name__ == "__main__":
    run_discovery_loop(epochs=100)
````

````python
import numpy as np
import scipy.linalg
from typing import List, Dict, Tuple

# ==========================================
# 1. Multi-Species Prime Valences
# ==========================================
# Instead of uniform entanglement, we force different elements 
# to use isolated sub-manifolds of the Arithmetic Vacuum.
SPECIES_PRIMES = {
    'Cu': [2, 3, 5, 7, 11],       # Copper uses low, dense primes
    'O':  [13, 17, 19, 23, 29],   # Oxygen uses higher, sparser primes
    'Ba': [31, 37, 41, 43, 47]    # Barium acts as an insulating spacer
}

class CuprateTensorOracle:
    def _map_to_fock_exponents(self, intensity: float, species: str, seed_idx: int) -> Dict[int, int]:
        """Maps interaction intensity strictly within the species' prime sub-manifold."""
        np.random.seed(seed_idx + int(intensity * 1000) % 1000)
        allowed_primes = SPECIES_PRIMES[species]
        num_active = min(int(intensity * len(allowed_primes)), len(allowed_primes))
        
        exponents = {p: 0 for p in allowed_primes}
        if num_active > 0:
            chosen_primes = np.random.choice(allowed_primes, num_active, replace=False)
            for p in chosen_primes:
                exponents[p] = int(np.random.randint(1, 4))
        return exponents

    def evaluate_lattice(self, atoms: List[Dict], stability_threshold: float) -> Dict[str, float]:
        num_atoms = len(atoms)
        positions = np.array([a['pos'] for a in atoms])
        species_list = [a['species'] for a in atoms]
        
        # Background Metric H_0 (Gaussian RBF)
        dist_matrix = np.linalg.norm(positions[:, None, :] - positions[None, :, :], axis=-1)
        H_0 = np.exp(-dist_matrix**2 / 2.0)
        np.fill_diagonal(H_0, 1.0)
        H_0 += np.eye(num_atoms) * 1e-5
        
        # Map to Species-Specific Arithmetic Fock Space
        atom_exponents = []
        for i in range(num_atoms):
            local_density = np.sum(H_0[i]) - 1.0
            atom_exponents.append(self._map_to_fock_exponents(local_density, species_list[i], i))
            
        # Interaction Graph H_int
        H_int = np.zeros((num_atoms, num_atoms))
        raw_bonds = 0
        
        for i in range(num_atoms):
            for j in range(i + 1, num_atoms):
                shared_log_val = 0.0
                exp_i = atom_exponents[i]
                exp_j = atom_exponents[j]
                
                # Cross-species entanglement only occurs if they are forced 
                # into a topological boundary condition (simulated here by thermal mixing)
                common_primes = set(exp_i.keys()).intersection(set(exp_j.keys()))
                
                # If they don't share primes natively, we check for spatial hybridization
                if not common_primes and dist_matrix[i, j] < 1.5:
                    shared_log_val += np.log(3) * (1.5 - dist_matrix[i, j])
                
                for p in common_primes:
                    if exp_i[p] > 0 and exp_j[p] > 0:
                        shared_log_val += min(exp_i[p], exp_j[p]) * np.log(p)
                
                if shared_log_val > 0:
                    H_int[i, j] = shared_log_val
                    H_int[j, i] = shared_log_val
                    raw_bonds += 1

        # Apply Obstruction-Filtered Positivity Sheaf
        coupling_constant = 0.08
        H_int_scaled = H_int * coupling_constant
        try:
            eigenvalues = scipy.linalg.eigh(H_int_scaled, H_0, eigvals_only=True)
            max_eig = np.max(np.abs(eigenvalues))
        except:
            max_eig = 10.0
            
        stable_bonds = raw_bonds
        if max_eig > 1.0:
            stable_bonds = int(raw_bonds * (1.0 - ((1.0 - 1.0/max_eig) * stability_threshold)))

        # ==========================================
        # 3. The 2D Stripe Reward Signal
        # ==========================================
        # We explicitly reward planar entanglement (dz ~ 0) and penalize 3D spherical blobs
        planar_bonds = 0
        vertical_bonds = 0
        
        for i in range(num_atoms):
            for j in range(i + 1, num_atoms):
                if H_int[i, j] > 0:
                    dz = abs(positions[i, 2] - positions[j, 2])
                    if dz < 0.5:
                        planar_bonds += 1
                    else:
                        vertical_bonds += 1
        
        # High Tc requires strong planar correlation and weak vertical correlation
        stripe_ratio = (planar_bonds + 1e-5) / (vertical_bonds + 1.0)
        coordination = 1.0 / (np.std(np.sum(H_int > 0, axis=1)) + 1.0)
        
        sc_score = (stable_bonds / (raw_bonds + 1e-5)) * stripe_ratio * coordination * 150.0
        
        return {
            "max_instability": max_eig,
            "stripe_ratio": stripe_ratio,
            "tc_score": min(sc_score, 135.0) # Cap at typical YBCO max
        }


# ==========================================
# 2. Lattice Constraints (Generator)
# ==========================================
class CuprateLatticeGenerator:
    def __init__(self, grid_size: int = 2):
        self.atoms = []
        # Initialize a structured, layered 2D grid
        for x in range(grid_size):
            for y in range(grid_size):
                # Layer 1: Copper-Oxygen Plane
                self.atoms.append({'species': 'Cu', 'pos': np.array([x*2.0, y*2.0, 0.0])})
                self.atoms.append({'species': 'O',  'pos': np.array([x*2.0 + 1.0, y*2.0, 0.0])})
                # Layer 2: Insulating Barium Spacer
                self.atoms.append({'species': 'Ba', 'pos': np.array([x*2.0 + 0.5, y*2.0 + 0.5, 2.0])})

    def propose_mutation(self, temperature: float) -> List[Dict]:
        proposed = []
        for atom in self.atoms:
            mutated_pos = atom['pos'] + np.random.normal(0.0, temperature, 3)
            # Lattice Restoring Force: Pull atoms back toward planar layers (z=0 or z=2)
            target_z = 0.0 if atom['species'] in ['Cu', 'O'] else 2.0
            mutated_pos[2] = mutated_pos[2] - 0.2 * (mutated_pos[2] - target_z)
            
            proposed.append({'species': atom['species'], 'pos': mutated_pos})
        return proposed

    def update_lattice(self, new_atoms: List[Dict]):
        self.atoms = new_atoms

# ==========================================
# Execution Loop
# ==========================================
def run_cuprate_search():
    oracle = CuprateTensorOracle()
    generator = CuprateLatticeGenerator(grid_size=2)
    
    best_tc = 0.0
    mutation_temp = 0.5
    
    print(f"{'Epoch':<6}{'Instability':<14}{'2D Stripe Ratio':<18}{'Est. Tc (K)':<12}")
    print("-" * 55)
    
    for epoch in range(1, 101):
        proposed = generator.propose_mutation(mutation_temp)
        metrics = oracle.evaluate_lattice(proposed, stability_threshold=0.85)
        
        tc = metrics["tc_score"]
        if tc > best_tc:
            best_tc = tc
            generator.update_lattice(proposed)
            
        mutation_temp = max(0.02, mutation_temp * 0.95)
        
        if epoch % 10 == 0:
            print(f"{epoch:<6}{metrics['max_instability']:<14.3f}{metrics['stripe_ratio']:<18.2f}{tc:<12.1f}")
            
    print("-" * 55)
    print(f"Discovery phase finished.")
    print(f"Optimal Cuprate Phase Transition: {best_tc:.2f} Kelvin")

if __name__ == "__main__":
    run_cuprate_search()
````

````python
import numpy as np
import scipy.linalg
from typing import List, Dict, Tuple

class PeriodicFockMapper:
    """
    Dynamically maps the entire Periodic Table into the Arithmetic Fock Space.
    Calculates subshell fillings via the Madelung Rule, applies quantum anomalies,
    and maps the active valence shells to the prime sequence.
    """
    def __init__(self, max_primes=1000):
        self.primes = self._sieve(max_primes)
        self.orbital_order, self.orbital_caps = self._generate_madelung_orbitals()
        
        # Rigorous mapping of orbital energy levels to the prime sequence
        self.orbital_to_prime = {orb: self.primes[i] for i, orb in enumerate(self.orbital_order)}
        
        # Standard Periodic Table (Z = index + 1)
        self.symbols = [
            "H", "He", "Li", "Be", "B", "C", "N", "O", "F", "Ne", "Na", "Mg", "Al", "Si", "P", "S", "Cl", "Ar",
            "K", "Ca", "Sc", "Ti", "V", "Cr", "Mn", "Fe", "Co", "Ni", "Cu", "Zn", "Ga", "Ge", "As", "Se", "Br", "Kr",
            "Rb", "Sr", "Y", "Zr", "Nb", "Mo", "Tc", "Ru", "Rh", "Pd", "Ag", "Cd", "In", "Sn", "Sb", "Te", "I", "Xe",
            "Cs", "Ba", "La", "Ce", "Pr", "Nd", "Pm", "Sm", "Eu", "Gd", "Tb", "Dy", "Ho", "Er", "Tm", "Yb", "Lu",
            "Hf", "Ta", "W", "Re", "Os", "Ir", "Pt", "Au", "Hg", "Tl", "Pb", "Bi", "Po", "At", "Rn"
        ]
        self.symbol_to_Z = {sym: z + 1 for z, sym in enumerate(self.symbols)}
        
        # Hardcoded overrides for Standard Model quantum anomalies (where electrons drop to lower orbitals)
        self.anomalies = {
            24: {'4s': 1, '3d': 5},    # Chromium
            29: {'4s': 1, '3d': 10},   # Copper
            42: {'5s': 1, '4d': 5},    # Molybdenum
            47: {'5s': 1, '4d': 10},   # Silver
            79: {'6s': 1, '5d': 10}    # Gold
        }

    def _sieve(self, limit: int) -> List[int]:
        sieve = [True] * limit
        for p in range(2, int(limit**0.5) + 1):
            if sieve[p]:
                for i in range(p * p, limit, p): sieve[i] = False
        return [p for p in range(2, limit) if sieve[p]]

    def _generate_madelung_orbitals(self) -> Tuple[List[str], Dict[str, int]]:
        """Generates orbital filling order sorted by (n + l), then n."""
        subshells = []
        l_map = {'s': 0, 'p': 1, 'd': 2, 'f': 3}
        caps = {'s': 2, 'p': 6, 'd': 10, 'f': 14}
        
        for n in range(1, 8):
            for l_char in ['s', 'p', 'd', 'f']:
                l = l_map[l_char]
                if l < n:  # Valid subshells only (e.g., no 1p, 2d)
                    subshells.append({'name': f"{n}{l_char}", 'n': n, 'l': l, 'energy': n + l})
                    
        # Sort by n+l, then by n
        subshells.sort(key=lambda x: (x['energy'], x['n']))
        order = [s['name'] for s in subshells]
        return order, caps

    def _get_electron_config(self, symbol: str) -> Dict[str, int]:
        """Fills orbitals up to Z electrons."""
        Z = self.symbol_to_Z.get(symbol)
        if not Z: raise ValueError(f"Unknown element: {symbol}")
        
        if Z in self.anomalies:
            # Reconstruct from anomaly template
            config = {}
            for orb in self.orbital_order:
                config[orb] = self.orbital_caps[orb[-1]] # Fill everything
                if orb in self.anomalies[Z]: break       # Stop at anomaly layer
            config.update(self.anomalies[Z])
            return {k: v for k, v in config.items() if v > 0}

        remaining = Z
        config = {}
        for orb in self.orbital_order:
            if remaining <= 0: break
            cap = self.orbital_caps[orb[-1]]
            fill = min(remaining, cap)
            config[orb] = fill
            remaining -= fill
            
        return config

    def get_fock_signature(self, symbol: str) -> Dict[int, int]:
        """
        Filters for VALENCE electrons (highest n, plus incomplete d/f shells)
        and returns the rigorously mapped Prime Arithmetic signature.
        """
        config = self._get_electron_config(symbol)
        
        # Determine highest principal quantum number (n)
        max_n = max(int(orb[0]) for orb in config.keys())
        
        valence_config = {}
        for orb, electrons in config.items():
            n = int(orb[0])
            l = orb[1]
            cap = self.orbital_caps[l]
            
            # Keep if it is the outermost shell (s, p)
            if n == max_n:
                valence_config[orb] = electrons
            # Keep if it is a partially filled inner shell (d, f), OR full but active (n-1 d)
            elif l in ['d', 'f'] and (n == max_n - 1 or n == max_n - 2):
                 valence_config[orb] = electrons
                 
        # Translate to Prime Dimensions
        prime_signature = {}
        for orb, electrons in valence_config.items():
            p = self.orbital_to_prime[orb]
            prime_signature[p] = electrons
            
        return prime_signature
````

````python
import torch
import torch.nn as nn
import numpy as np
from typing import List, Dict

# Automatically utilize GPU if available (for the Cluster)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Divergence Lab Engine initialized on: {DEVICE}")

class BatchedTensorOracle(nn.Module):
    """
    The PyTorch-accelerated Arithmetic Tensor Oracle.
    Evaluates thousands of atomic configurations simultaneously.
    """
    def __init__(self, num_atoms: int, max_primes: int = 50):
        super().__init__()
        self.num_atoms = num_atoms
        self.primes = torch.tensor(self._sieve(max_primes), dtype=torch.float32, device=DEVICE)
        self.log_primes = torch.log(self.primes)
        self.hybridization_idx = -1 # Use the largest prime in the set for vacuum hybridization

    def _sieve(self, limit: int) -> List[int]:
        sieve = [True] * limit
        for p in range(2, int(limit**0.5) + 1):
            if sieve[p]:
                for i in range(p * p, limit, p): sieve[i] = False
        return [p for p in range(2, limit) if sieve[p]]

    def forward(self, positions: torch.Tensor, valences: torch.Tensor, stability_threshold: float = 0.85):
        """
        positions: (Batch, Atoms, 3) - The spatial coordinates.
        valences:  (Batch, Atoms, Num_Primes) - The programmatic Fock signatures.
        """
        B, N, _ = positions.shape

        # ---------------------------------------------------------
        # 1. Upgrade H_0: Screened Coulomb Metric (Yukawa Potential)
        # ---------------------------------------------------------
        # Calculate pairwise distances (Batch, Atoms, Atoms)
        dist_matrix = torch.cdist(positions, positions)
        
        # Screened Coulomb: e^(-kappa * r) / (r + epsilon)
        # We add a strong diagonal ridge to guarantee positive-definiteness for Cholesky
        kappa = 1.2 # Screening length
        H_0 = torch.exp(-kappa * dist_matrix) / (dist_matrix + 0.1)
        
        # Zero out the diagonal from the cdist, and replace with a strict identity ridge
        eye = torch.eye(N, device=DEVICE).unsqueeze(0).expand(B, N, N)
        H_0 = H_0 * (1.0 - eye) + eye * 5.0 

        # ---------------------------------------------------------
        # 2. Vectorized Arithmetic Entanglement (H_int)
        # ---------------------------------------------------------
        # valences: (B, N, P). We want min(val[i], val[j]) for all pairs.
        # Broadcast to (B, N, 1, P) and (B, 1, N, P) -> (B, N, N, P)
        v_i = valences.unsqueeze(2)
        v_j = valences.unsqueeze(1)
        shared_valences = torch.minimum(v_i, v_j)

        # Dot product with log(primes) over the P dimension -> (B, N, N)
        H_int = torch.sum(shared_valences * self.log_primes, dim=-1)
        
        # Mask out self-interaction
        H_int = H_int * (1.0 - eye)

        # ---------------------------------------------------------
        # 3. Dynamic Geometric Hybridization (PyTorch Vectorized)
        # ---------------------------------------------------------
        # If distance < 2.5 and dz < 0.2, trigger the hybridization prime bond
        dz = torch.abs(positions[:, :, 2].unsqueeze(2) - positions[:, :, 2].unsqueeze(1))
        hybrid_mask = (dist_matrix < 2.5) & (dz < 0.2) & (1.0 - eye).bool()
        
        hybrid_intensity = ((2.5 - dist_matrix) * 3.0).clamp(min=0.0) * hybrid_mask
        H_int += hybrid_intensity * self.log_primes[self.hybridization_idx]

        # ---------------------------------------------------------
        # 4. The Obstruction-Filtered Positivity Sheaf (Batched Eigendecomposition)
        # ---------------------------------------------------------
        coupling_constant = 0.015
        H_int_scaled = H_int * coupling_constant

        # Explicitly construct H_0^(-1/2) via eigendecomposition and clamp the spectrum
        D, V = torch.linalg.eigh(H_0)
        D = torch.clamp(D, min=1e-4) 
        D_inv_sqrt = torch.diag_embed(1.0 / torch.sqrt(D))
        H_0_inv_half = torch.bmm(torch.bmm(V, D_inv_sqrt), V.transpose(1, 2))
        
        # Transform the interaction matrix: C = H_0^(-1/2) @ H_int_scaled @ H_0^(-1/2)
        C = torch.bmm(torch.bmm(H_0_inv_half, H_int_scaled), H_0_inv_half)
        
        # Compute the final eigenvalues of the stabilized system
        eigenvalues = torch.linalg.eigvalsh(C)
        max_eig = torch.max(torch.abs(eigenvalues), dim=1).values

        # ---------------------------------------------------------
        # 5. Differentiable Reward Signal (Soft Masking)
        # ---------------------------------------------------------
        # We replace hard booleans with steep sigmoids (k=20.0) 
        # so PyTorch Autograd can calculate continuous spatial gradients.
        k = 20.0
        soft_bond = torch.sigmoid(k * (H_int - 0.01)) 
        
        raw_bonds = torch.sum(soft_bond, dim=(1, 2)) / 2.0
        
        instability_penalty = torch.clamp(1.0 - (1.0 / max_eig), min=0.0)
        survival_rate = 1.0 - (instability_penalty * stability_threshold)
        stable_bonds = raw_bonds * survival_rate

        # Soft 2D Stripe Reward
        dz = torch.abs(positions[:, :, 2].unsqueeze(2) - positions[:, :, 2].unsqueeze(1))
        soft_planar = torch.sigmoid(k * (0.3 - dz)) 
        
        planar_mask = soft_planar * soft_bond
        vertical_mask = (1.0 - soft_planar) * soft_bond
        
        planar_bonds = torch.sum(planar_mask, dim=(1, 2)) / 2.0
        vertical_bonds = torch.sum(vertical_mask, dim=(1, 2)) / 2.0
        
        stripe_ratio = (planar_bonds + 1e-4) / (vertical_bonds + 1.0)
        
        # Calculate raw Tc score
        raw_tc = (stable_bonds / (raw_bonds + 1e-4)) * stripe_ratio * 50.0
        capped_tc = torch.clamp(raw_tc, max=135.0)

        # The Non-Negotiable Stability Gate
        # If max_eig > 1.0, the reward is dynamically crushed.
        stability_gate = torch.sigmoid(10.0 * (1.05 - max_eig))
        tc_score = capped_tc * stability_gate

        return tc_score, max_eig, stripe_ratio


class LangevinDiffusionGenerator:
    """
    Replaces the random walk. Uses gradients from the Oracle to 
    simulate Score-Based Diffusion over the crystal lattice coordinates.
    """
    def __init__(self, batch_size: int, num_atoms: int):
        self.batch_size = batch_size
        self.num_atoms = num_atoms
        
        # Initialize random positions with requires_grad=True to enable Autograd
        self.positions = nn.Parameter(
            torch.randn(batch_size, num_atoms, 3, device=DEVICE)
        )
        # Adam optimizer acts as our Langevin dynamics step
        self.optimizer = torch.optim.Adam([self.positions], lr=0.1)

    def generate_static_valences(self) -> torch.Tensor:
        """
        Simulates the output of the PeriodicFockMapper.
        We assign alternating Cu and O signatures to the batch.
        """
        # Shape: (Batch, Atoms, Primes)
        valences = torch.zeros((self.batch_size, self.num_atoms, 15), device=DEVICE)
        
        # Copper (Primes index 5, 6)
        valences[:, 0::2, 5] = 1.0
        valences[:, 0::2, 6] = 10.0
        
        # Oxygen (Prime index 2)
        valences[:, 1::2, 2] = 4.0
        
        return valences

    def step(self, oracle: BatchedTensorOracle) -> torch.Tensor:
        self.optimizer.zero_grad()
        valences = self.generate_static_valences()
        
        tc_scores, max_eigs, stripe_ratios = oracle(self.positions, valences)
        
        # ---------------------------------------------------------
        # Physical Geometry & Stability Penalties
        # ---------------------------------------------------------
        # 1. Instability Loss (Directly penalize supercritical states)
        instability_loss = torch.mean(torch.relu(max_eigs - 1.0) ** 2)
        
        # 2. Distance Constraints
        dist_matrix = torch.cdist(self.positions, self.positions)
        B, N, _ = self.positions.shape
        eye = torch.eye(N, device=DEVICE).unsqueeze(0).expand(B, N, N)
        dist_matrix = dist_matrix + eye * 10.0 # Push diagonal out of min_dist penalty range
        
        # Prevent atoms from overlapping (fusion) or flying too far apart (evaporation)
        min_dist_penalty = torch.mean(torch.relu(1.2 - dist_matrix) ** 2)
        max_dist_penalty = torch.mean(torch.relu(dist_matrix - 4.0) ** 2)
        
        # 3. Prevent degenerate 2D collapse (ensure realistic layer spacing)
        z_collapse_penalty = torch.relu(0.15 - torch.std(self.positions[:, :, 2], dim=1)).mean()
        
        # 4. Center of mass dispersion
        center = torch.mean(self.positions, dim=1, keepdim=True)
        dispersion_penalty = torch.mean((self.positions - center)**2, dim=(1, 2))
        
        # ---------------------------------------------------------
        # Total Loss Function
        # ---------------------------------------------------------
        loss = (-torch.mean(tc_scores) 
                + (15.0 * instability_loss) 
                + (5.0 * min_dist_penalty) 
                + (2.0 * max_dist_penalty) 
                + (5.0 * z_collapse_penalty) 
                + (0.5 * torch.mean(dispersion_penalty)))
        
        loss.backward()
        
        with torch.no_grad():
            noise = torch.randn_like(self.positions) * 0.05
            self.positions.grad += noise
            
        self.optimizer.step()
        
        return tc_scores, max_eigs, stripe_ratios

# ==========================================
# Execution: Workload Configurable Loop
# ==========================================
def run_production_cluster(workload: str = "laptop"):
    # Configure workload
    if workload == "laptop":
        BATCH_SIZE = 16
        print("Configuring for local compute: Batched 16 parallel simulations.")
    elif workload == "cluster":
        BATCH_SIZE = 1024
        print("Configuring for cluster compute: Batched 1024 parallel simulations.")

    NUM_ATOMS = 12
    EPOCHS = 100

    oracle = BatchedTensorOracle(num_atoms=NUM_ATOMS).to(DEVICE)
    generator = LangevinDiffusionGenerator(batch_size=BATCH_SIZE, num_atoms=NUM_ATOMS)

    print(f"\n{'Epoch':<8}{'Instability':<15}{'2D Stripe Ratio':<20}{'Est. Tc (K)':<15}")
    print("-" * 60)

    for epoch in range(1, EPOCHS + 1):
        tc_scores, max_eigs, stripe_ratios = generator.step(oracle)
        
        if epoch % 10 == 0:
            # Find the best performing crystal in the current batch
            best_idx = torch.argmax(tc_scores)
            best_tc = tc_scores[best_idx].item()
            best_stripe = stripe_ratios[best_idx].item()
            
            # Instability is monitored as a batch average to ensure overall physics stability
            mean_instability = torch.mean(max_eigs).item()
            
            print(f"{epoch:<8}{mean_instability:<15.3f}{best_stripe:<20.2f}{best_tc:<15.1f}")

    print("-" * 60)
    print(f"Discovery phase finished.")
    
if __name__ == "__main__":
    # Change to "cluster" if running on a GPU rig
    run_production_cluster(workload="laptop")
````

````python
import torch
import torch.nn as nn
import numpy as np
from typing import List, Dict, Tuple

# =========================================================================
# 0. Hardware Config
# =========================================================================
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Divergence Lab Engine initialized on: {DEVICE}")

# =========================================================================
# 1. The Programmatic Periodic-Fock Mapper
# =========================================================================
class PeriodicFockMapper:
    """Dynamically derives Arithmetic Fock signatures from Standard Model quantum numbers."""
    def __init__(self, max_primes: int = 50):
        self.primes = self._sieve(1000)[:max_primes] # Get first 'max_primes' primes
        self.prime_tensor = torch.tensor(self.primes, dtype=torch.float32, device=DEVICE)
        self.orbital_order, self.orbital_caps = self._generate_madelung_orbitals()
        self.orbital_to_prime_idx = {orb: i for i, orb in enumerate(self.orbital_order)}
        
        self.symbols = [
            "H", "He", "Li", "Be", "B", "C", "N", "O", "F", "Ne", "Na", "Mg", "Al", "Si", "P", "S", "Cl", "Ar",
            "K", "Ca", "Sc", "Ti", "V", "Cr", "Mn", "Fe", "Co", "Ni", "Cu", "Zn", "Ga", "Ge", "As", "Se", "Br", "Kr",
            "Rb", "Sr", "Y", "Zr", "Nb", "Mo", "Tc", "Ru", "Rh", "Pd", "Ag", "Cd", "In", "Sn", "Sb", "Te", "I", "Xe",
            "Cs", "Ba" # Extending up to Barium for YBCO
        ]
        self.symbol_to_Z = {sym: z + 1 for z, sym in enumerate(self.symbols)}
        
        # Quantum anomalies where electrons jump shells to maximize stability
        self.anomalies = {
            24: {'4s': 1, '3d': 5},    # Chromium
            29: {'4s': 1, '3d': 10},   # Copper
        }

    def _sieve(self, limit: int) -> List[int]:
        sieve = [True] * limit
        for p in range(2, int(limit**0.5) + 1):
            if sieve[p]:
                for i in range(p * p, limit, p): sieve[i] = False
        return [p for p in range(2, limit) if sieve[p]]

    def _generate_madelung_orbitals(self) -> Tuple[List[str], Dict[str, int]]:
        subshells = []
        l_map = {'s': 0, 'p': 1, 'd': 2, 'f': 3}
        caps = {'s': 2, 'p': 6, 'd': 10, 'f': 14}
        for n in range(1, 8):
            for l_char in ['s', 'p', 'd', 'f']:
                l = l_map[l_char]
                if l < n:
                    subshells.append({'name': f"{n}{l_char}", 'n': n, 'l': l, 'energy': n + l})
        subshells.sort(key=lambda x: (x['energy'], x['n']))
        return [s['name'] for s in subshells], caps

    def _get_valence_config(self, symbol: str) -> Dict[str, int]:
        Z = self.symbol_to_Z.get(symbol)
        if not Z: raise ValueError(f"Unknown element: {symbol}")
        
        # Fill orbitals
        config = {}
        if Z in self.anomalies:
            for orb in self.orbital_order:
                config[orb] = self.orbital_caps[orb[-1]]
                if orb in self.anomalies[Z]: break
            config.update(self.anomalies[Z])
        else:
            remaining = Z
            for orb in self.orbital_order:
                if remaining <= 0: break
                fill = min(remaining, self.orbital_caps[orb[-1]])
                config[orb] = fill
                remaining -= fill
                
        # Filter for active valences (highest n, and incomplete inner d/f shells)
        max_n = max(int(orb[0]) for orb in config.keys() if config[orb] > 0)
        valence_config = {}
        for orb, electrons in config.items():
            if electrons == 0: continue
            n, l = int(orb[0]), orb[1]
            if n == max_n or (l in ['d', 'f'] and n >= max_n - 2):
                 valence_config[orb] = electrons
        return valence_config

    def build_tensor_for_atoms(self, symbols: List[str]) -> torch.Tensor:
        """Returns a dense PyTorch tensor of shape (Num_Atoms, Max_Primes)"""
        num_atoms = len(symbols)
        num_primes = len(self.primes)
        valences = torch.zeros((num_atoms, num_primes), dtype=torch.float32, device=DEVICE)
        
        for i, sym in enumerate(symbols):
            config = _get_valence_config = self._get_valence_config(sym)
            for orb, electrons in config.items():
                idx = self.orbital_to_prime_idx[orb]
                valences[i, idx] = electrons
        return valences

# =========================================================================
# 2. The Differentiable Batched Oracle (Physics Engine)
# =========================================================================
class BatchedTensorOracle(nn.Module):
    def __init__(self, mapper: PeriodicFockMapper):
        super().__init__()
        self.primes = mapper.prime_tensor
        self.log_primes = torch.log(self.primes)
        self.num_primes = len(self.primes)
        
        # We reserve the highest prime in the set for vacuum hybridization
        self.hybrid_prime_idx = self.num_primes - 1 

    def forward(self, positions: torch.Tensor, valences: torch.Tensor, stability_threshold: float = 0.85):
        B, N, _ = positions.shape
        
        # 1. Screened Coulomb Metric (H_0)
        dist_matrix = torch.cdist(positions, positions)
        kappa = 1.2
        eye = torch.eye(N, device=DEVICE).unsqueeze(0).expand(B, N, N)
        H_0 = torch.exp(-kappa * dist_matrix) / (dist_matrix + 0.1)
        H_0 = H_0 * (1.0 - eye) + eye * 5.0 

        # 2. Arithmetic Entanglement (H_int)
        v_i = valences.unsqueeze(2) # (B, N, 1, P)
        v_j = valences.unsqueeze(1) # (B, 1, N, P)
        shared_valences = torch.minimum(v_i, v_j)
        H_int = torch.sum(shared_valences * self.log_primes, dim=-1) * (1.0 - eye)

        # 3. Dynamic Geometric Hybridization
        dz = torch.abs(positions[:, :, 2].unsqueeze(2) - positions[:, :, 2].unsqueeze(1))
        hybrid_mask = (dist_matrix < 2.5) & (dz < 0.2) & (1.0 - eye).bool()
        hybrid_intensity = ((2.5 - dist_matrix) * 3.0).clamp(min=0.0) * hybrid_mask
        H_int += hybrid_intensity * self.log_primes[self.hybrid_prime_idx]

        # 4. Obstruction-Filtered Positivity Sheaf (Shock-Absorbed)
        coupling_constant = 0.015  # Adjusted scale to allow optimization headroom
        H_int_scaled = H_int * coupling_constant

        D, V = torch.linalg.eigh(H_0)
        D = torch.clamp(D, min=1e-4) 
        D_inv_sqrt = torch.diag_embed(1.0 / torch.sqrt(D))
        H_0_inv_half = torch.bmm(torch.bmm(V, D_inv_sqrt), V.transpose(1, 2))
        
        C = torch.bmm(torch.bmm(H_0_inv_half, H_int_scaled), H_0_inv_half)
        eigenvalues = torch.linalg.eigvalsh(C)
        max_eig = torch.max(torch.abs(eigenvalues), dim=1).values

        # 5. Differentiable Reward Signal (Soft Masking)
        k = 20.0
        soft_bond = torch.sigmoid(k * (H_int - 0.01)) 
        raw_bonds = torch.sum(soft_bond, dim=(1, 2)) / 2.0
        
        instability_penalty = torch.clamp(1.0 - (1.0 / max_eig), min=0.0)
        survival_rate = 1.0 - (instability_penalty * stability_threshold)
        stable_bonds = raw_bonds * survival_rate

        soft_planar = torch.sigmoid(k * (0.3 - dz)) 
        planar_mask = soft_planar * soft_bond
        vertical_mask = (1.0 - soft_planar) * soft_bond
        
        planar_bonds = torch.sum(planar_mask, dim=(1, 2)) / 2.0
        vertical_bonds = torch.sum(vertical_mask, dim=(1, 2)) / 2.0
        
        stripe_ratio = (planar_bonds + 1e-4) / (vertical_bonds + 1.0)
        
        raw_tc = (stable_bonds / (raw_bonds + 1e-4)) * stripe_ratio * 50.0
        capped_tc = torch.clamp(raw_tc, max=135.0)

        # Non-Negotiable Stability Gate
        stability_gate = torch.sigmoid(10.0 * (1.05 - max_eig))
        tc_score = capped_tc * stability_gate
        # We also apply the stability gate to the raw score so we can see 
        # the true unbounded reward of stable structures.
        raw_tc_gated = raw_tc * stability_gate

        return tc_score, raw_tc_gated, max_eig, stripe_ratio

# =========================================================================
# 3. The Chemically-Typed Local-Coordination Unit Cell Generator (v4.3)
# =========================================================================
class CuprateUnitCellGenerator:
    def __init__(self, batch_size: int, mapper: PeriodicFockMapper):
        self.batch_size = batch_size

        # YBa2Cu3O7 stoichiometry: 13 atoms
        self.symbols = (
            ['Cu'] * 3 +
            ['O'] * 7 +
            ['Y'] * 1 +
            ['Ba'] * 2
        )
        self.num_atoms = len(self.symbols)

        # Useful typed indices
        self.cu_indices = [i for i, s in enumerate(self.symbols) if s == "Cu"]
        self.o_indices = [i for i, s in enumerate(self.symbols) if s == "O"]
        self.y_indices = [i for i, s in enumerate(self.symbols) if s == "Y"]
        self.ba_indices = [i for i, s in enumerate(self.symbols) if s == "Ba"]

        # Build arithmetic Fock valence tensor
        base_valences = mapper.build_tensor_for_atoms(self.symbols)
        self.valences = base_valences.unsqueeze(0).expand(batch_size, -1, -1)

        # Species masks
        self.cu_o_mask = self._build_species_mask("Cu", "O").unsqueeze(0).expand(batch_size, -1, -1)
        # Additional typed repulsion masks
        self.o_o_mask = self._build_same_species_mask("O").unsqueeze(0).expand(batch_size, -1, -1)
        self.cu_cu_mask = self._build_same_species_mask("Cu").unsqueeze(0).expand(batch_size, -1, -1)
        self.ba_ba_mask = self._build_same_species_mask("Ba").unsqueeze(0).expand(batch_size, -1, -1)
        self.y_ba_mask = self._build_species_mask("Y", "Ba").unsqueeze(0).expand(batch_size, -1, -1)
        self.nonself_mask = (1.0 - torch.eye(self.num_atoms, device=DEVICE)).unsqueeze(0).expand(batch_size, -1, -1)

        # Layered initialization
        initial_positions = torch.zeros((batch_size, self.num_atoms, 3), device=DEVICE)

        for idx, sym in enumerate(self.symbols):
            initial_positions[:, idx, :2] = torch.randn(batch_size, 2, device=DEVICE) * 1.5

            if sym in ["Cu", "O"]:
                # Conducting Cu-O plane
                initial_positions[:, idx, 2] = torch.randn(batch_size, device=DEVICE) * 0.08
            elif sym == "Y":
                # Spacer / charge reservoir layer
                initial_positions[:, idx, 2] = 1.8 + torch.randn(batch_size, device=DEVICE) * 0.08
            elif sym == "Ba":
                # Spacer / reservoir layer
                initial_positions[:, idx, 2] = 2.1 + torch.randn(batch_size, device=DEVICE) * 0.08

        self.positions = nn.Parameter(initial_positions)
        self.optimizer = torch.optim.Adam([self.positions], lr=0.08)

    def _build_species_mask(self, species_a: str, species_b: str) -> torch.Tensor:
        """
        Directed pairwise mask for interactions of type species_a-species_b.
        Includes both A->B and B->A directions.
        """
        mask = torch.zeros((self.num_atoms, self.num_atoms), device=DEVICE)

        for i, si in enumerate(self.symbols):
            for j, sj in enumerate(self.symbols):
                if i == j:
                    continue

                if (si == species_a and sj == species_b) or (si == species_b and sj == species_a):
                    mask[i, j] = 1.0

        return mask
    
    def _build_same_species_mask(self, species: str) -> torch.Tensor:
        """
        Creates a directed pairwise mask for same-species repulsion,
        excluding self-interactions.
        """
        mask = torch.zeros((self.num_atoms, self.num_atoms), device=DEVICE)

        for i, si in enumerate(self.symbols):
            for j, sj in enumerate(self.symbols):
                if i != j and si == species and sj == species:
                    mask[i, j] = 1.0

        return mask

    def _local_cu_o_window_penalty(
        self,
        dist_matrix: torch.Tensor,
        lower: float = 1.75,
        upper: float = 2.15,
        k_nearest: int = 4,
    ) -> torch.Tensor:
        """
        Enforce that each Cu has a local shell of nearby O atoms inside the
        chemically plausible Cu-O bond window.

        This avoids the bad v4.2 behavior where every Cu-O pair was forced
        into the bond window.
        """
        penalties = []

        # Only allow Cu-O distances. Everything else is made artificially large.
        cu_o_dist = dist_matrix + (1.0 - self.cu_o_mask) * 100.0

        actual_k = min(k_nearest, len(self.o_indices))

        for cu_idx in self.cu_indices:
            # Shape: (B, N)
            d_i = cu_o_dist[:, cu_idx, :]

            # Nearest O atoms to this Cu
            nearest_dists, _ = torch.topk(
                d_i,
                k=actual_k,
                largest=False,
                dim=1
            )

            # Penalize too-short and too-long local Cu-O bonds
            short_barrier = torch.relu(1.70 - nearest_dists) ** 4

            penalty_i = (
                15.0 * short_barrier +
                torch.relu(lower - nearest_dists) ** 2 +
                torch.relu(nearest_dists - upper) ** 2
            ).mean(dim=1)

            penalties.append(penalty_i)

        return torch.stack(penalties, dim=1).mean()

    def _cu_coordination_penalty(
        self,
        dist_matrix: torch.Tensor,
        lower: float = 1.75,
        upper: float = 2.15,
        target_coordination: float = 4.0,
        sharpness: float = 20.0,
    ) -> torch.Tensor:
        """
        Softly encourages each Cu to have approximately target_coordination
        oxygen neighbors in the Cu-O bond window.
        """
        cu_o_dist = dist_matrix + (1.0 - self.cu_o_mask) * 100.0

        # Smooth bond indicator: high only when lower < d < upper
        bond_strength = (
            torch.sigmoid(sharpness * (cu_o_dist - lower)) *
            torch.sigmoid(sharpness * (upper - cu_o_dist)) *
            self.cu_o_mask
        )

        # Count O-neighbors for each Cu row
        cu_coord = bond_strength[:, self.cu_indices, :].sum(dim=2)

        return torch.mean((cu_coord - target_coordination) ** 2)

    def _layer_regularization(self) -> torch.Tensor:
        """
        Encourages Cu/O to remain near the conducting plane and Y/Ba to remain
        in spacer layers, without forcing total collapse.
        """
        z = self.positions[:, :, 2]

        cu_o_indices = self.cu_indices + self.o_indices
        spacer_indices = self.y_indices + self.ba_indices

        cu_o_z = z[:, cu_o_indices]
        spacer_z = z[:, spacer_indices]

        # Cu-O plane centered near z=0
        cu_o_plane_penalty = torch.mean(cu_o_z ** 2)

        # Spacers should remain separated from Cu-O plane
        spacer_height_penalty = torch.mean(torch.relu(1.4 - spacer_z) ** 2)

        # Prevent all z-values from collapsing into a single sheet
        z_spread_penalty = torch.relu(0.35 - torch.std(z, dim=1)).mean()

        return cu_o_plane_penalty + spacer_height_penalty + z_spread_penalty

    def step(self, oracle: BatchedTensorOracle) -> Tuple[torch.Tensor, ...]:
        self.optimizer.zero_grad()

        tc_scores, raw_tc_scores, max_eigs, stripe_ratios = oracle(
            self.positions,
            self.valences
        )

        # ---------------------------------------------------------
        # 1. Cone-contractivity / topological stability
        # ---------------------------------------------------------
        instability_loss = torch.mean(torch.relu(max_eigs - 1.0) ** 2)

        # ---------------------------------------------------------
        # 2. Pairwise distances
        # ---------------------------------------------------------
        dist_matrix = torch.cdist(self.positions, self.positions)

        # Ignore self-distances for geometric penalties
        safe_dist = dist_matrix + (1.0 - self.nonself_mask) * 10.0

        # ---------------------------------------------------------
        # 3. Global geometry constraints
        # ---------------------------------------------------------
        min_dist_penalty = torch.sum(
            self.nonself_mask * torch.relu(1.15 - safe_dist) ** 2
        ) / (torch.sum(self.nonself_mask) + 1e-6)

        # Replace all-pairs max distance penalty with soft bounding box.
        # This avoids forcing the whole structure into an artificial ball.
        xy_bound_penalty = torch.mean(torch.relu(torch.abs(self.positions[:, :, :2]) - 3.5) ** 2)
        z_bound_penalty = torch.mean(torch.relu(torch.abs(self.positions[:, :, 2]) - 3.0) ** 2)

        center = torch.mean(self.positions, dim=1, keepdim=True)
        dispersion_penalty = torch.mean((self.positions - center) ** 2)

        # ---------------------------------------------------------
        # Species-resolved short-range repulsion
        # ---------------------------------------------------------
        o_o_repulsion = torch.sum(
            self.o_o_mask * torch.relu(1.60 - dist_matrix) ** 2
        ) / (torch.sum(self.o_o_mask) + 1e-6)

        # Strong local penalty against the closest O-O collapse
        o_o_dist = dist_matrix + (1.0 - self.o_o_mask) * 100.0
        nearest_o_o_dist, _ = torch.min(o_o_dist, dim=2)

        # Only evaluate rows corresponding to oxygen atoms
        nearest_o_o_for_oxygen = nearest_o_o_dist[:, self.o_indices]

        o_o_nearest_penalty = torch.mean(
            torch.relu(1.40 - nearest_o_o_for_oxygen) ** 2
        )
        cu_cu_repulsion = torch.sum(
            self.cu_cu_mask * torch.relu(2.20 - dist_matrix) ** 2
        ) / (torch.sum(self.cu_cu_mask) + 1e-6)

        ba_ba_repulsion = torch.sum(
            self.ba_ba_mask * torch.relu(2.20 - dist_matrix) ** 2
        ) / (torch.sum(self.ba_ba_mask) + 1e-6)

        y_ba_repulsion = torch.sum(
            self.y_ba_mask * torch.relu(1.80 - dist_matrix) ** 2
        ) / (torch.sum(self.y_ba_mask) + 1e-6)

        # Strong local penalty against Cu-Cu collapse
        cu_cu_dist = dist_matrix + (1.0 - self.cu_cu_mask) * 100.0
        nearest_cu_cu_dist, _ = torch.min(cu_cu_dist, dim=2)
        nearest_cu_cu_for_cu = nearest_cu_cu_dist[:, self.cu_indices]
        nearest_cu_cu_value = torch.min(nearest_cu_cu_dist).item()

        cu_cu_nearest_penalty = torch.mean(
            torch.relu(2.40 - nearest_cu_cu_for_cu) ** 2
        )

        typed_repulsion_penalty = (
            120.0 * o_o_repulsion +
            150.0 * o_o_nearest_penalty +
            18.0 * cu_cu_repulsion +
            200.0 * cu_cu_nearest_penalty +
            8.0 * ba_ba_repulsion +
            8.0 * y_ba_repulsion
        )

        # ---------------------------------------------------------
        # 4. Chemically typed local Cu-O constraints
        # ---------------------------------------------------------
        cu_o_window_penalty = self._local_cu_o_window_penalty(
            dist_matrix,
            lower=1.75,
            upper=2.15,
            k_nearest=4
        )

        cu_coord_penalty = self._cu_coordination_penalty(
            dist_matrix,
            lower=1.75,
            upper=2.15,
            target_coordination=4.0
        )

        # ---------------------------------------------------------
        # 5. Layer structure regularization
        # ---------------------------------------------------------
        layer_penalty = self._layer_regularization()

        # ---------------------------------------------------------
        # 6. Combined loss
        # ---------------------------------------------------------
        loss = (
            -torch.mean(tc_scores)
            + 15.0 * instability_loss
            + 30.0 * min_dist_penalty
            + 40.0 * cu_o_window_penalty
            + 8.0 * cu_coord_penalty
            + 4.0 * layer_penalty
            + 1.0 * xy_bound_penalty
            + 1.0 * z_bound_penalty
            + 0.3 * dispersion_penalty
            + typed_repulsion_penalty
        )

        loss.backward()

        # Langevin noise
        with torch.no_grad():
            self.positions.grad += torch.randn_like(self.positions) * 0.025

        self.optimizer.step()

        return tc_scores, raw_tc_scores, max_eigs, stripe_ratios, o_o_repulsion, torch.min(nearest_cu_cu_dist)

# =========================================================================
# 4. Execution Loop
# =========================================================================
def run_divergence_lab():
    BATCH_SIZE = 32
    EPOCHS = 300
    
    mapper = PeriodicFockMapper(max_primes=50)
    oracle = BatchedTensorOracle(mapper=mapper).to(DEVICE)
    generator = CuprateUnitCellGenerator(batch_size=BATCH_SIZE, mapper=mapper)
    
    print("\n--- DIVERGENCE LAB v4.1: UNBOUNDED STOICHIOMETRIC OPTIMIZATION ---")
    print(f"Targeting: 13-Atom YBCO Unit Cell (YBa2Cu3O7)")
    print(f"{'Epoch':<8}{'Instability':<15}{'Stripe Ratio':<15}{'Capped Tc':<15}{'Raw Tc':<15}{'o_o_repulsion':<15}{'nearest_cu_cu':<15}")
    print("-" * 70)

    for epoch in range(1, EPOCHS + 1):
        tc_scores, raw_tc_scores, max_eigs, stripe_ratios, o_o_repulsion, nearest_cu_cu = generator.step(oracle)        
        if epoch % 10 == 0:
            # We track the best lattice based on the RAW score to see the true ceiling
            best_idx = torch.argmax(raw_tc_scores)
            best_capped_tc = tc_scores[best_idx].item()
            best_raw_tc = raw_tc_scores[best_idx].item()
            best_stripe = stripe_ratios[best_idx].item()
            mean_instability = torch.mean(max_eigs).item()
            print(f"{epoch:<8}{mean_instability:<15.3f}{best_stripe:<15.2f}{best_capped_tc:<15.1f}{best_raw_tc:<15.1f}{o_o_repulsion.item():<15.4f}{nearest_cu_cu.item():<15.3f}")

    print("-" * 70)
    print("Discovery phase finished.")
    
    # ---------------------------------------------------------
    # Final Geometry Dump
    # ---------------------------------------------------------
    best_idx = torch.argmax(raw_tc_scores)
    best_positions = generator.positions[best_idx].detach().cpu().numpy()
    symbols = generator.symbols
    
    print("\n--- FINAL OPTIMIZED GEOMETRY (Best Batch Element) ---")
    print(f"Final Raw Tc Score: {raw_tc_scores[best_idx].item():.2f}")
    print(f"{'Atom':<6}{'X':<10}{'Y':<10}{'Z':<10}")
    print("-" * 35)
    
    for sym, pos in zip(symbols, best_positions):
        print(f"{sym:<6}{pos[0]:>8.3f}  {pos[1]:>8.3f}  {pos[2]:>8.3f}")

if __name__ == "__main__":
    run_divergence_lab()
````

````python
import torch
import torch.nn as nn
import numpy as np
from typing import List, Dict, Tuple, Optional

# =========================================================================
# 0. Hardware Config
# =========================================================================
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Divergence Lab Engine initialized on: {DEVICE}")

# =========================================================================
# 1. The Programmatic Periodic-Fock Mapper
# =========================================================================
class PeriodicFockMapper:
    """Dynamically derives Arithmetic Fock signatures from Standard Model quantum numbers."""
    def __init__(self, max_primes: int = 50):
        self.primes = self._sieve(1000)[:max_primes]
        self.prime_tensor = torch.tensor(self.primes, dtype=torch.float32, device=DEVICE)
        self.orbital_order, self.orbital_caps = self._generate_madelung_orbitals()
        self.orbital_to_prime_idx = {orb: i for i, orb in enumerate(self.orbital_order)}

        self.symbols = [
            "H","He","Li","Be","B","C","N","O","F","Ne","Na","Mg","Al","Si","P","S","Cl","Ar",
            "K","Ca","Sc","Ti","V","Cr","Mn","Fe","Co","Ni","Cu","Zn","Ga","Ge","As","Se","Br","Kr",
            "Rb","Sr","Y","Zr","Nb","Mo","Tc","Ru","Rh","Pd","Ag","Cd","In","Sn","Sb","Te","I","Xe",
            "Cs","Ba"
        ]
        self.symbol_to_Z = {sym: z + 1 for z, sym in enumerate(self.symbols)}
        self.anomalies = {
            24: {'4s': 1, '3d': 5},
            29: {'4s': 1, '3d': 10},
        }

    def _sieve(self, limit: int) -> List[int]:
        sieve = [True] * limit
        for p in range(2, int(limit**0.5) + 1):
            if sieve[p]:
                for i in range(p * p, limit, p): sieve[i] = False
        return [p for p in range(2, limit) if sieve[p]]

    def _generate_madelung_orbitals(self) -> Tuple[List[str], Dict[str, int]]:
        subshells = []
        l_map = {'s': 0, 'p': 1, 'd': 2, 'f': 3}
        caps = {'s': 2, 'p': 6, 'd': 10, 'f': 14}
        for n in range(1, 8):
            for l_char in ['s', 'p', 'd', 'f']:
                l = l_map[l_char]
                if l < n:
                    subshells.append({'name': f"{n}{l_char}", 'n': n, 'l': l, 'energy': n + l})
        subshells.sort(key=lambda x: (x['energy'], x['n']))
        return [s['name'] for s in subshells], caps

    def _get_valence_config(self, symbol: str) -> Dict[str, int]:
        Z = self.symbol_to_Z.get(symbol)
        if not Z: raise ValueError(f"Unknown element: {symbol}")
        config = {}
        if Z in self.anomalies:
            for orb in self.orbital_order:
                config[orb] = self.orbital_caps[orb[-1]]
                if orb in self.anomalies[Z]: break
            config.update(self.anomalies[Z])
        else:
            remaining = Z
            for orb in self.orbital_order:
                if remaining <= 0: break
                fill = min(remaining, self.orbital_caps[orb[-1]])
                config[orb] = fill
                remaining -= fill
        max_n = max(int(orb[0]) for orb in config.keys() if config[orb] > 0)
        valence_config = {}
        for orb, electrons in config.items():
            if electrons == 0: continue
            n, l = int(orb[0]), orb[1]
            if n == max_n or (l in ['d', 'f'] and n >= max_n - 2):
                valence_config[orb] = electrons
        return valence_config

    def build_tensor_for_atoms(self, symbols: List[str]) -> torch.Tensor:
        num_atoms = len(symbols)
        num_primes = len(self.primes)
        valences = torch.zeros((num_atoms, num_primes), dtype=torch.float32, device=DEVICE)
        for i, sym in enumerate(symbols):
            config = self._get_valence_config(sym)
            for orb, electrons in config.items():
                idx = self.orbital_to_prime_idx[orb]
                valences[i, idx] = electrons
        return valences


# =========================================================================
# 2. Adaptive Lagrange Multiplier Governor
# =========================================================================
class LagrangeGovernor:
    """
    Self-adjusting constraint weighting via dual ascent.

    Each constraint lambda evolves as:
        lambda_i <- clamp(lambda_i * decay + lr * violation_i, min_val, max_val)

    Persistent violations automatically attract larger penalties.
    Satisfied constraints stop growing and slowly decay.
    """
    def __init__(
        self,
        constraint_names: List[str],
        init_values: Dict[str, float],
        lr: float = 2.0,
        decay: float = 0.995,
        max_val: float = 600.0,
        min_val: float = 0.5,
    ):
        self.lr = lr
        self.decay = decay
        self.max_val = max_val
        self.min_val = min_val

        self.lambdas: Dict[str, torch.Tensor] = {
            name: torch.tensor(init_values.get(name, 10.0), dtype=torch.float32, device=DEVICE)
            for name in constraint_names
        }
        # History for diagnostics
        self.history: Dict[str, List[float]] = {name: [] for name in constraint_names}

    def loss(self, violations: Dict[str, torch.Tensor]) -> torch.Tensor:
        """Weighted sum of violations using current lambdas."""
        total = torch.tensor(0.0, device=DEVICE)
        for name, v in violations.items():
            total = total + self.lambdas[name] * v
        return total

    @torch.no_grad()
    def step(self, violations: Dict[str, torch.Tensor]):
        """Dual ascent: increase lambda where violations persist, decay where satisfied."""
        for name, v in violations.items():
            violation_scalar = v.detach().clamp(min=0.0)
            self.lambdas[name] = torch.clamp(
                self.lambdas[name] * self.decay + self.lr * violation_scalar,
                min=self.min_val,
                max=self.max_val,
            )
            self.history[name].append(self.lambdas[name].item())

    def report(self) -> str:
        parts = [f"  {name}: λ={val.item():.1f}" for name, val in self.lambdas.items()]
        return "\n".join(parts)


# =========================================================================
# 3. The Differentiable Batched Oracle (Physics Engine)  — unchanged
# =========================================================================
class BatchedTensorOracle(nn.Module):
    def __init__(self, mapper: PeriodicFockMapper):
        super().__init__()
        self.primes = mapper.prime_tensor
        self.log_primes = torch.log(self.primes)
        self.num_primes = len(self.primes)
        self.hybrid_prime_idx = self.num_primes - 1

    def forward(self, positions: torch.Tensor, valences: torch.Tensor, stability_threshold: float = 0.85):
        B, N, _ = positions.shape
        dist_matrix = torch.cdist(positions, positions)
        kappa = 1.2
        eye = torch.eye(N, device=DEVICE).unsqueeze(0).expand(B, N, N)

        # ---------------------------------------------------------------
        # H_0: Pure screened Coulomb metric — kept clean.
        #
        # v5.3: Pauli/physical exclusion is enforced ONLY in the
        # constraint penalty layer, never in H_0.  Mixing repulsion
        # into H_0 inflates C-matrix eigenvalues → max_eig >> 1 →
        # stability gate permanently closed → zero Tc score.
        # The oracle's job is to score bonding topology; geometry
        # feasibility is the constraint layer's job.
        # ---------------------------------------------------------------
        coulomb = torch.exp(-kappa * dist_matrix) / (dist_matrix + 0.1)
        H_0 = coulomb * (1.0 - eye) + eye * 5.0

        # ---------------------------------------------------------------
        # H_int: Arithmetic entanglement gated by a physical bond window.
        #
        # Previous version: H_int was nonzero at ANY distance, so two atoms
        # at 0.5 Å still produced full bonding reward.
        #
        # v5.1: multiply by a smooth gate that is 1 inside (d_bond_min,
        # d_bond_max) and falls to 0 outside.  No bond below 1.40 Å;
        # no bond above 3.00 Å.  The gate is differentiable everywhere.
        # ---------------------------------------------------------------
        d_bond_min = 1.40    # Å — hard lower physical limit for any bond
        d_bond_max = 3.00    # Å — upper cutoff (includes longer Y/Ba bonds)
        k_gate = 8.0   # softer gate: gradient doesn't spike at d_bond_min
        bond_gate = (
            torch.sigmoid(k_gate * (dist_matrix - d_bond_min)) *
            torch.sigmoid(k_gate * (d_bond_max - dist_matrix))
        ) * (1.0 - eye)

        v_i = valences.unsqueeze(2)
        v_j = valences.unsqueeze(1)
        shared_valences = torch.minimum(v_i, v_j)
        H_int = torch.sum(shared_valences * self.log_primes, dim=-1) * bond_gate

        dz = torch.abs(positions[:, :, 2].unsqueeze(2) - positions[:, :, 2].unsqueeze(1))
        hybrid_mask = (dist_matrix < 2.5) & (dz < 0.2) & bond_gate.bool()
        hybrid_intensity = ((2.5 - dist_matrix) * 3.0).clamp(min=0.0) * hybrid_mask
        H_int += hybrid_intensity * self.log_primes[self.hybrid_prime_idx]

        coupling_constant = 0.015
        H_int_scaled = H_int.clamp(max=50.0) * coupling_constant  # clamp prevents NaN in eigh

        D, V = torch.linalg.eigh(H_0)
        D = torch.clamp(D, min=1e-4)
        D_inv_sqrt = torch.diag_embed(1.0 / torch.sqrt(D))
        H_0_inv_half = torch.bmm(torch.bmm(V, D_inv_sqrt), V.transpose(1, 2))

        C = torch.bmm(torch.bmm(H_0_inv_half, H_int_scaled), H_0_inv_half)
        eigenvalues = torch.linalg.eigvalsh(C)
        max_eig = torch.max(torch.abs(eigenvalues), dim=1).values

        k = 20.0
        soft_bond = torch.sigmoid(k * (H_int - 0.01))
        raw_bonds = torch.sum(soft_bond, dim=(1, 2)) / 2.0

        instability_penalty = torch.clamp(1.0 - (1.0 / max_eig), min=0.0)
        survival_rate = 1.0 - (instability_penalty * stability_threshold)
        stable_bonds = raw_bonds * survival_rate

        soft_planar = torch.sigmoid(k * (0.3 - dz))
        planar_mask = soft_planar * soft_bond
        vertical_mask = (1.0 - soft_planar) * soft_bond

        planar_bonds = torch.sum(planar_mask, dim=(1, 2)) / 2.0
        vertical_bonds = torch.sum(vertical_mask, dim=(1, 2)) / 2.0

        stripe_ratio = (planar_bonds + 1e-4) / (vertical_bonds + 1.0)
        raw_tc = (stable_bonds / (raw_bonds + 1e-4)) * stripe_ratio * 50.0
        capped_tc = torch.clamp(raw_tc, max=135.0)

        stability_gate = torch.sigmoid(10.0 * (1.05 - max_eig))
        tc_score = capped_tc * stability_gate
        raw_tc_gated = raw_tc * stability_gate

        return tc_score, raw_tc_gated, max_eig, stripe_ratio


# =========================================================================
# 4. YBCO Unit Cell Generator v5 — Adaptive Lagrangian
# =========================================================================
class CuprateUnitCellGeneratorV5:
    """
    v5: replaces all hand-tuned penalty coefficients with an adaptive
    Lagrange multiplier governor (dual ascent).  Constraint violations
    are normalized to dimensionless fractional-violation^2 units so
    that the governor operates on a common scale.

    Fixed-scale terms kept outside the governor:
      - layer_penalty (structural regularizer, not a hard chemical constraint)
      - xy/z bounding box (soft wall, not chemistry)
      - dispersion (compactness prior)
    """
    def __init__(self, batch_size: int, mapper: PeriodicFockMapper):
        self.batch_size = batch_size

        self.symbols = ['Cu'] * 3 + ['O'] * 7 + ['Y'] * 1 + ['Ba'] * 2
        self.num_atoms = len(self.symbols)

        self.cu_indices = [i for i, s in enumerate(self.symbols) if s == "Cu"]
        self.o_indices  = [i for i, s in enumerate(self.symbols) if s == "O"]
        self.y_indices  = [i for i, s in enumerate(self.symbols) if s == "Y"]
        self.ba_indices = [i for i, s in enumerate(self.symbols) if s == "Ba"]

        base_valences = mapper.build_tensor_for_atoms(self.symbols)
        self.valences = base_valences.unsqueeze(0).expand(batch_size, -1, -1)

        self.cu_o_mask    = self._build_species_mask("Cu", "O").unsqueeze(0).expand(batch_size, -1, -1)
        self.o_o_mask     = self._build_same_species_mask("O").unsqueeze(0).expand(batch_size, -1, -1)
        self.cu_cu_mask   = self._build_same_species_mask("Cu").unsqueeze(0).expand(batch_size, -1, -1)
        self.ba_ba_mask   = self._build_same_species_mask("Ba").unsqueeze(0).expand(batch_size, -1, -1)
        self.y_ba_mask    = self._build_species_mask("Y", "Ba").unsqueeze(0).expand(batch_size, -1, -1)
        self.nonself_mask = (1.0 - torch.eye(self.num_atoms, device=DEVICE)).unsqueeze(0).expand(batch_size, -1, -1)

        # ---- Structured initialization: place atoms on a grid with guaranteed
        #      minimum separation, then add small random noise.
        #      This prevents the Pauli wall from seeing collapsed geometries
        #      before the optimizer has had a chance to react.
        # ---------------------------------------------------------------
        # Grid layout: Cu and O share the xy plane, packed on a ~2.5 Å grid.
        # Y and Ba are in their spacer layers.
        cu_o_xy_grid = [
            # Cu positions
            [0.0,  0.0], [2.6,  0.0], [1.3,  2.25],
            # O positions (bridging)
            [1.3,  0.0], [0.65, 1.12], [1.95, 1.12],
            [0.0,  2.25],[2.6,  2.25], [3.25, 1.12],
            [1.3, -1.12],
        ]
        # Center and expand for batch
        initial_positions = torch.zeros((batch_size, self.num_atoms, 3), device=DEVICE)

        cu_o_count = len(self.cu_indices) + len(self.o_indices)  # 10
        for local_idx, global_idx in enumerate(self.cu_indices + self.o_indices):
            xy = cu_o_xy_grid[local_idx % len(cu_o_xy_grid)]
            initial_positions[:, global_idx, 0] = xy[0] + torch.randn(batch_size, device=DEVICE) * 0.15
            initial_positions[:, global_idx, 1] = xy[1] + torch.randn(batch_size, device=DEVICE) * 0.15
            initial_positions[:, global_idx, 2] =          torch.randn(batch_size, device=DEVICE) * 0.05

        # Y: center of the cell, spacer layer
        for global_idx in self.y_indices:
            initial_positions[:, global_idx, 0] = 1.3  + torch.randn(batch_size, device=DEVICE) * 0.15
            initial_positions[:, global_idx, 1] = 1.12 + torch.randn(batch_size, device=DEVICE) * 0.15
            initial_positions[:, global_idx, 2] = 1.85 + torch.randn(batch_size, device=DEVICE) * 0.05

        # Ba: two Ba at different z-heights to seed the asymmetry constraint
        ba_xy = [[0.0, 0.0], [2.6, 2.25]]
        ba_z  = [1.95, 2.25]
        for k, global_idx in enumerate(self.ba_indices):
            initial_positions[:, global_idx, 0] = ba_xy[k][0] + torch.randn(batch_size, device=DEVICE) * 0.15
            initial_positions[:, global_idx, 1] = ba_xy[k][1] + torch.randn(batch_size, device=DEVICE) * 0.15
            initial_positions[:, global_idx, 2] = ba_z[k]     + torch.randn(batch_size, device=DEVICE) * 0.05

        self.positions = nn.Parameter(initial_positions)
        self.optimizer = torch.optim.Adam([self.positions], lr=0.08)

        # ---- Lagrange governor ----
        # Initial values chosen to approximate the v4 weights but now
        # all constraints are normalized to dimensionless units, so these
        # are calibrated starting points, not magic numbers.
        self.governor = LagrangeGovernor(
            constraint_names=[
                "instability",
                "global_min_dist",
                "o_o_repulsion",
                "o_o_nearest",
                "cu_cu_repulsion",
                "cu_cu_nearest",
                "ba_ba_repulsion",
                "y_ba_repulsion",
                "cu_o_window",
                "cu_coord",
                "ba_asymmetry",   # NEW: breaks Ba z-symmetry
            ],
            init_values={
                "instability":    8.0,   # lower init: linear penalty needs less scale
                "global_min_dist": 20.0,
                "o_o_repulsion":   40.0,
                "o_o_nearest":     50.0,
                "cu_cu_repulsion": 20.0,
                "cu_cu_nearest":   60.0,
                "ba_ba_repulsion": 8.0,
                "y_ba_repulsion":  8.0,
                "cu_o_window":     40.0,
                "cu_coord":        10.0,
                "ba_asymmetry":    5.0,
            },
            lr=1.5,
            decay=0.997,
            max_val=600.0,
            min_val=0.5,
        )

        # Reference distances for normalization
        self.d_ref = {
            "global_min": 1.15,
            "o_o":        1.60,
            "o_o_near":   1.40,
            "cu_cu":      2.20,
            "cu_cu_near": 2.50,  # raised from 2.40 to push Cu-Cu to physical range
            "ba_ba":      2.20,
            "y_ba":       1.80,
            "cu_o_low":   1.75,
            "cu_o_high":  2.15,
        }

    # ------------------------------------------------------------------
    # Species masks
    # ------------------------------------------------------------------
    def _build_species_mask(self, a: str, b: str) -> torch.Tensor:
        mask = torch.zeros((self.num_atoms, self.num_atoms), device=DEVICE)
        for i, si in enumerate(self.symbols):
            for j, sj in enumerate(self.symbols):
                if i != j and ((si == a and sj == b) or (si == b and sj == a)):
                    mask[i, j] = 1.0
        return mask

    def _build_same_species_mask(self, sp: str) -> torch.Tensor:
        mask = torch.zeros((self.num_atoms, self.num_atoms), device=DEVICE)
        for i, si in enumerate(self.symbols):
            for j, sj in enumerate(self.symbols):
                if i != j and si == sp and sj == sp:
                    mask[i, j] = 1.0
        return mask

    # ------------------------------------------------------------------
    # Normalized constraints (dimensionless fractional-violation^2)
    # ------------------------------------------------------------------
    def _violations(self, dist_matrix: torch.Tensor) -> Dict[str, torch.Tensor]:
        d = self.d_ref
        safe_dist = dist_matrix + (1.0 - self.nonself_mask) * 10.0

        # --- Global minimum distance (hard physical exclusion wall) ---
        # Use a softplus-based wall so gradients are always finite.
        # This is the ONLY place physical repulsion enters the system.
        # alpha=5, beta=8, onset=1.15 Å → strong repulsion below 1.15 Å,
        # zero above it, smooth everywhere.
        alpha_ex = 5.0
        beta_ex  = 8.0
        onset_ex = 1.15  # Å — below any real chemical bond
        exclusion_wall = alpha_ex * torch.nn.functional.softplus(
            beta_ex * (onset_ex - safe_dist)
        ) * self.nonself_mask
        global_min_dist = exclusion_wall.sum() / (self.nonself_mask.sum() + 1e-6)

        # --- O-O repulsion ---
        frac_oo = torch.relu((d["o_o"] - dist_matrix) / d["o_o"]) ** 2
        o_o_repulsion = torch.sum(self.o_o_mask * frac_oo) / (torch.sum(self.o_o_mask) + 1e-6)

        # --- O-O nearest (local short-range) ---
        o_o_dist = dist_matrix + (1.0 - self.o_o_mask) * 100.0
        nearest_o_o_dist, _ = torch.min(o_o_dist, dim=2)
        nearest_o_o = nearest_o_o_dist[:, self.o_indices]
        frac_oo_near = torch.relu((d["o_o_near"] - nearest_o_o) / d["o_o_near"]) ** 2
        o_o_nearest = torch.mean(frac_oo_near)

        # --- Cu-Cu repulsion ---
        frac_cucu = torch.relu((d["cu_cu"] - dist_matrix) / d["cu_cu"]) ** 2
        cu_cu_repulsion = torch.sum(self.cu_cu_mask * frac_cucu) / (torch.sum(self.cu_cu_mask) + 1e-6)

        # --- Cu-Cu nearest (local hard floor) ---
        cu_cu_dist = dist_matrix + (1.0 - self.cu_cu_mask) * 100.0
        nearest_cu_cu_dist, _ = torch.min(cu_cu_dist, dim=2)
        nearest_cu_cu = nearest_cu_cu_dist[:, self.cu_indices]
        frac_cucu_near = torch.relu((d["cu_cu_near"] - nearest_cu_cu) / d["cu_cu_near"]) ** 2
        cu_cu_nearest = torch.mean(frac_cucu_near)

        # --- Ba-Ba repulsion ---
        frac_baba = torch.relu((d["ba_ba"] - dist_matrix) / d["ba_ba"]) ** 2
        ba_ba_repulsion = torch.sum(self.ba_ba_mask * frac_baba) / (torch.sum(self.ba_ba_mask) + 1e-6)

        # --- Y-Ba repulsion ---
        frac_yba = torch.relu((d["y_ba"] - dist_matrix) / d["y_ba"]) ** 2
        y_ba_repulsion = torch.sum(self.y_ba_mask * frac_yba) / (torch.sum(self.y_ba_mask) + 1e-6)

        # --- Cu-O window (local, k-nearest) ---
        cu_o_window = self._cu_o_window_normalized(dist_matrix)

        # --- Cu coordination ---
        cu_coord = self._cu_coordination_penalty(dist_matrix)

        # --- Ba z-asymmetry (NEW) ---
        ba_z = self.positions[:, self.ba_indices, 2]
        # Encourage |z_Ba1 - z_Ba2| >= 0.25 Å to break the symmetry
        ba_asymmetry = torch.mean(torch.relu(0.25 - torch.abs(ba_z[:, 0] - ba_z[:, 1])) ** 2)

        return {
            "global_min_dist": global_min_dist,
            "o_o_repulsion":   o_o_repulsion,
            "o_o_nearest":     o_o_nearest,
            "cu_cu_repulsion": cu_cu_repulsion,
            "cu_cu_nearest":   cu_cu_nearest,
            "ba_ba_repulsion": ba_ba_repulsion,
            "y_ba_repulsion":  y_ba_repulsion,
            "cu_o_window":     cu_o_window,
            "cu_coord":        cu_coord,
            "ba_asymmetry":    ba_asymmetry,
        }

    def _cu_o_window_normalized(
        self,
        dist_matrix: torch.Tensor,
        lower: float = 1.75,
        upper: float = 2.15,
        k_nearest: int = 4,
    ) -> torch.Tensor:
        cu_o_dist = dist_matrix + (1.0 - self.cu_o_mask) * 100.0
        actual_k = min(k_nearest, len(self.o_indices))
        penalties = []
        for cu_idx in self.cu_indices:
            d_i = cu_o_dist[:, cu_idx, :]
            nearest_dists, _ = torch.topk(d_i, k=actual_k, largest=False, dim=1)
            # Normalized: hard barrier below 1.70
            short_barrier = torch.relu((1.70 - nearest_dists) / 1.70) ** 4
            too_short = torch.relu((lower - nearest_dists) / lower) ** 2
            too_long  = torch.relu((nearest_dists - upper) / upper) ** 2
            penalty_i = (15.0 * short_barrier + too_short + too_long).mean(dim=1)
            penalties.append(penalty_i)
        return torch.stack(penalties, dim=1).mean()

    def _cu_coordination_penalty(
        self,
        dist_matrix: torch.Tensor,
        lower: float = 1.75,
        upper: float = 2.15,
        target_coordination: float = 4.0,
        sharpness: float = 20.0,
    ) -> torch.Tensor:
        cu_o_dist = dist_matrix + (1.0 - self.cu_o_mask) * 100.0
        bond_strength = (
            torch.sigmoid(sharpness * (cu_o_dist - lower)) *
            torch.sigmoid(sharpness * (upper - cu_o_dist)) *
            self.cu_o_mask
        )
        cu_coord = bond_strength[:, self.cu_indices, :].sum(dim=2)
        return torch.mean(((cu_coord - target_coordination) / target_coordination) ** 2)

    def _layer_regularization(self) -> torch.Tensor:
        z = self.positions[:, :, 2]
        cu_o_indices   = self.cu_indices + self.o_indices
        spacer_indices = self.y_indices + self.ba_indices
        cu_o_z   = z[:, cu_o_indices]
        spacer_z = z[:, spacer_indices]
        cu_o_plane_penalty    = torch.mean(cu_o_z ** 2)
        spacer_height_penalty = torch.mean(torch.relu(1.4 - spacer_z) ** 2)
        z_spread_penalty      = torch.relu(0.35 - torch.std(z, dim=1)).mean()
        return cu_o_plane_penalty + spacer_height_penalty + z_spread_penalty

    # ------------------------------------------------------------------
    # Main optimization step
    # ------------------------------------------------------------------
    def step(
        self,
        oracle: BatchedTensorOracle,
        epoch: int,
        noise_scale: float = 0.025,
    ) -> Tuple[torch.Tensor, ...]:
        self.optimizer.zero_grad()

        tc_scores, raw_tc_scores, max_eigs, stripe_ratios = oracle(
            self.positions, self.valences
        )

        dist_matrix = torch.cdist(self.positions, self.positions)

        # --- Instability constraint ---
        # Use linear penalty above 1.0 (not quadratic) so the governor
        # can pull it down without needing enormous lambda.
        instability_loss = torch.mean(torch.relu(max_eigs - 1.0))

        # --- All other normalized violations ---
        violations = self._violations(dist_matrix)
        violations["instability"] = instability_loss

        # --- Lagrangian constraint loss ---
        constraint_loss = self.governor.loss(violations)

        # --- Fixed-scale structural terms (not chemistry, not tuned) ---
        xy_bound   = torch.mean(torch.relu(torch.abs(self.positions[:, :, :2]) - 3.5) ** 2)
        z_bound    = torch.mean(torch.relu(torch.abs(self.positions[:, :, 2]) - 3.0) ** 2)
        center     = torch.mean(self.positions, dim=1, keepdim=True)
        dispersion = torch.mean((self.positions - center) ** 2)
        layer      = self._layer_regularization()

        loss = (
            -torch.mean(tc_scores)
            + constraint_loss
            + 4.0  * layer
            + 1.0  * xy_bound
            + 1.0  * z_bound
            + 0.3  * dispersion
        )

        loss.backward()

        # Guard: if any gradient is NaN (e.g. from a degenerate batch element),
        # zero it out so it doesn't poison the whole update.
        with torch.no_grad():
            if self.positions.grad is not None:
                nan_mask = torch.isnan(self.positions.grad)
                if nan_mask.any():
                    self.positions.grad[nan_mask] = 0.0

        # Gradient clipping: prevents the Pauli wall from producing a
        # single catastrophic step that launches atoms to infinity.
        torch.nn.utils.clip_grad_norm_([self.positions], max_norm=5.0)

        # Annealed Langevin noise
        with torch.no_grad():
            self.positions.grad += torch.randn_like(self.positions) * noise_scale

        self.optimizer.step()

        # --- Dual ascent: update lambdas AFTER gradient step ---
        self.governor.step(violations)

        cu_cu_dist  = dist_matrix + (1.0 - self.cu_cu_mask) * 100.0
        nearest_cu_cu, _ = torch.min(cu_cu_dist, dim=2)

        return (
            tc_scores,
            raw_tc_scores,
            max_eigs,
            stripe_ratios,
            violations["o_o_repulsion"],
            torch.min(nearest_cu_cu),
        )


# =========================================================================
# 5. Execution Loop
# =========================================================================
def run_divergence_lab_v5():
    BATCH_SIZE = 32
    EPOCHS     = 400

    # Noise schedule: full Langevin for exploration, then anneal for exploitation
    def noise_schedule(epoch: int) -> float:
        if epoch < 150:
            return 0.025
        elif epoch < 250:
            return 0.025 * (1.0 - (epoch - 150) / 200)   # linear decay
        else:
            return 0.0125  # low steady-state thermal noise

    mapper    = PeriodicFockMapper(max_primes=50)
    oracle    = BatchedTensorOracle(mapper=mapper).to(DEVICE)
    generator = CuprateUnitCellGeneratorV5(batch_size=BATCH_SIZE, mapper=mapper)

    print("\n--- DIVERGENCE LAB v5.3: CLEAN ORACLE + EXCLUSION IN CONSTRAINTS ---")
    print(f"Targeting: 13-Atom YBCO Unit Cell (YBa₂Cu₃O₇)")
    print(f"Oracle: clean Coulomb H_0 + bond-gated H_int [1.40–3.00Å] (Pauli in constraint layer)")
    print(f"Init:   structured Cu-O grid (min sep ~2.5Å) + seeded Ba z-asymmetry")
    print(f"Constraint governor: dual ascent, lr=1.5, decay=0.997, max_λ=600")
    print(f"\n{'Epoch':<8}{'Instab.':<12}{'Stripe':<10}{'CappedTc':<12}{'RawTc':<12}"
          f"{'OO-rep':<12}{'CuCu-min':<12}{'λ_cucu_n':<12}{'λ_oo_n':<12}")
    print("-" * 100)

    best_raw_ever = -np.inf
    best_geometry = None
    best_symbols  = generator.symbols
    MIN_VALID_CUCU  = 2.20   # Å — absolute minimum Cu-Cu for a valid geometry
    MIN_VALID_ANY   = 1.20   # Å — absolute minimum any-pair distance

    def _geometry_is_valid(positions_np: np.ndarray, cu_indices: List[int]) -> bool:
        """Return True only if the geometry passes hard physical distance checks."""
        pos = torch.tensor(positions_np, device='cpu')
        dist = torch.cdist(pos.unsqueeze(0), pos.unsqueeze(0)).squeeze(0)
        dist.fill_diagonal_(999.0)
        # Global floor
        if dist.min().item() < MIN_VALID_ANY:
            return False
        # Cu-Cu floor
        cu_pos = pos[cu_indices]
        cu_dist = torch.cdist(cu_pos.unsqueeze(0), cu_pos.unsqueeze(0)).squeeze(0)
        cu_dist.fill_diagonal_(999.0)
        if cu_dist.min().item() < MIN_VALID_CUCU:
            return False
        return True

    for epoch in range(1, EPOCHS + 1):
        noise = noise_schedule(epoch)
        tc_scores, raw_tc_scores, max_eigs, stripe_ratios, o_o_rep, nearest_cu_cu = \
            generator.step(oracle, epoch, noise_scale=noise)

        # Track global best — only accept geometrically valid structures
        # Sort candidates by score (descending) and take the first valid one
        sorted_idx = torch.argsort(raw_tc_scores, descending=True)
        for candidate_idx in sorted_idx:
            candidate_pos = generator.positions[candidate_idx].detach().cpu().numpy()
            if _geometry_is_valid(candidate_pos, generator.cu_indices):
                if raw_tc_scores[candidate_idx].item() > best_raw_ever:
                    best_raw_ever = raw_tc_scores[candidate_idx].item()
                    best_geometry = candidate_pos.copy()
                break  # only need the best valid one per epoch
        best_idx = sorted_idx[0]  # for display (may be invalid — that's OK to show)

        if epoch % 10 == 0:
            if torch.isnan(tc_scores).any():
                print(f"{epoch:<8}  *** NaN detected — check initialization or gradient scale ***")
                break
            best_capped = tc_scores[best_idx].item()
            best_raw    = raw_tc_scores[best_idx].item()
            best_stripe = stripe_ratios[best_idx].item()
            mean_instab = torch.mean(max_eigs).item()
            lam_cucu_n  = generator.governor.lambdas["cu_cu_nearest"].item()
            lam_oo_n    = generator.governor.lambdas["o_o_nearest"].item()

            print(f"{epoch:<8}{mean_instab:<12.3f}{best_stripe:<10.2f}"
                  f"{best_capped:<12.1f}{best_raw:<12.1f}"
                  f"{o_o_rep.item():<12.4f}{nearest_cu_cu.item():<12.3f}"
                  f"{lam_cucu_n:<12.1f}{lam_oo_n:<12.1f}")

    print("-" * 100)

    # Lambda evolution summary
    print("\n--- FINAL LAGRANGE MULTIPLIERS ---")
    print(generator.governor.report())

    # ---- Final geometry dump ----
    if best_geometry is None:
        print("\nWARNING: No geometrically valid structure found across all epochs.")
        print("Consider raising MIN_VALID_CUCU or running more epochs.")
        return

    print("\n--- FINAL OPTIMIZED GEOMETRY (Global Best — Validated) ---")
    print(f"Best Raw Tc Score (all epochs): {best_raw_ever:.2f}")
    print(f"\n{'Atom':<6}{'X':>10}{'Y':>10}{'Z':>10}")
    print("-" * 38)
    for sym, pos in zip(best_symbols, best_geometry):
        print(f"{sym:<6}{pos[0]:>10.3f}{pos[1]:>10.3f}{pos[2]:>10.3f}")

    # ---- Geometry quality checks ----
    print("\n--- GEOMETRY QUALITY REPORT ---")
    pos = torch.tensor(best_geometry, device=DEVICE)
    dist = torch.cdist(pos.unsqueeze(0), pos.unsqueeze(0)).squeeze(0)

    cu_pos = pos[generator.cu_indices]
    o_pos  = pos[generator.o_indices]
    ba_pos = pos[generator.ba_indices]

    # Cu-O distances
    cu_o_dists = torch.cdist(cu_pos, o_pos)
    print(f"\nCu-O distances (nearest 4 per Cu, target 1.75–2.15 Å):")
    for i, cu_idx in enumerate(generator.cu_indices):
        d_row = cu_o_dists[i]
        nearest4, _ = torch.topk(d_row, k=4, largest=False)
        vals = " ".join([f"{v.item():.3f}" for v in nearest4])
        print(f"  Cu[{cu_idx}]: {vals}")

    # Cu-Cu distances
    cu_cu_dists = torch.cdist(cu_pos, cu_pos)
    cu_cu_dists.fill_diagonal_(999.0)
    min_cucu = cu_cu_dists.min().item()
    print(f"\nNearest Cu-Cu distance: {min_cucu:.3f} Å  (target ≥ 2.50 Å)")

    # Ba z-separation
    if len(generator.ba_indices) >= 2:
        ba_dz = abs(ba_pos[0, 2].item() - ba_pos[1, 2].item())
        print(f"Ba z-separation: {ba_dz:.3f} Å  (target ≥ 0.25 Å)")

    # Cu-O plane flatness
    cu_o_z = pos[generator.cu_indices + generator.o_indices, 2]
    print(f"Cu/O z-spread (std): {cu_o_z.std().item():.4f} Å  (target < 0.05 Å)")


if __name__ == "__main__":
    run_divergence_lab_v5()
````
